```python
import os
import re
from datetime import datetime, timedelta

directory = '/Users/xiehao/Desktop/workspace/X2AI/FinAI/WebServer/app/static/reports'
md_files = [f for f in os.listdir(directory) if f.endswith('.markdown')]

date_pattern = re.compile(r'(\d{8})')  # 匹配8位数字的日期
dated_reports = {}
for file in md_files:
    match = date_pattern.search(file)
    if match:
        date_str = match.group(1)
        try:
            with open(f'{directory}/{file}', 'r', encoding='utf-8') as file:
                content = file.read()
            dated_reports[date_str] = content
        except ValueError:
            continue  # 如果日期格式无效，跳过该文件
dated_reports
```
